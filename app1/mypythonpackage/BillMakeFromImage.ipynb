```python
import pytesseract
from PIL import Image
import cv2
import os
import numpy as np

# Path to Tesseract-OCR executable (set according to your installation)
pytesseract.pytesseract.tesseract_cmd = r"D:/Downloads/tesseract-ocr-w64-setup-5.4.0.20240606.exe"  # Change this for your OS
img = "D:/getPaytmPaymentRecords/paymentImages/billAug.jpg"
image_directory = "D:/getPaytmPaymentRecords/paymentImages/"


def preprocess_image(image_path):
    """Preprocess image for better OCR accuracy."""
    img = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    img = cv2.threshold(img, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)[1]
    return img

def extract_text(image_path):
    """Extract text using Tesseract OCR."""
    img = preprocess_image(image_path)
    text = pytesseract.image_to_string(img)
    return text

def main(directory):
    """Process multiple images in a directory."""
    extracted_data = []
    for filename in os.listdir(directory):
        if filename.endswith((".png", ".jpg", ".jpeg")):
            image_path = os.path.join(directory, filename)
            text = extract_text(image_path)
            extracted_data.append((filename, text))
    
    # Save results
    with open("extracted_texts.txt", "w", encoding="utf-8") as f:
        for file, text in extracted_data:
            f.write(f"{file}:/n{text}\n{'-'*50}\n")

main(image_directory)
```

```python
from PIL import Image
import pytesseract
import re

# Path to the image
image_path = 'D:/getPaytmPaymentRecords/paymentImages/billAug.jpg'

# Open the image
image = Image.open(image_path)

# Get image dimensions
width, height = image.size

# Split the image into two parts
upper_half = image.crop((0, 0, width, height // 2))
lower_half = image.crop((0, height // 2, width, height))

# Save the split images (optional)
upper_half.save('D:/getPaytmPaymentRecords/paymentImages/upper_half.jpg')
lower_half.save('D:/getPaytmPaymentRecords/paymentImages/lower_half.jpg')

# Perform OCR on both parts
text_upper = pytesseract.image_to_string(upper_half)
text_lower = pytesseract.image_to_string(lower_half)

# Combine the extracted text
extracted_text = text_upper + '\n' + text_lower

# Print all extracted text
print("Extracted Text:")
print(extracted_text)
# Extract data using regex 
pattern = re.compile(r'(\w[\w\s]+)\s*-\s*₹?\s*([\d,.]+)\s+Paid on\s+([\d]+\s+\w+,\s+[\d:]+\s*[APM]{2})', re.IGNORECASE)

records = pattern.findall(extracted_text)

# Write the data in tabular form
output_path = 'D:/getPaytmPaymentRecords/paymentImages/extracted_text.txt'

with open(output_path, 'w') as f:
    f.write(f"{'Name':<30}{'Price':<10}{'Date':<20}\n")
    f.write("-" * 60 + "\n")
    for record in records:
        name, price, date = record
        f.write(f"{name:<30}{price:<10}{date:<20}\n")

print("Text extraction and formatting completed. Extracted text saved to 'extracted_text.txt'")
```
